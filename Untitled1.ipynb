```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
import torch.nn as nn

# ===== CONFIG =====
DEVICE = "mps" if torch.backends.mps.is_available() else "cpu"
IMG_SIZE = 256
MODEL_PATH = "best_unet.pth"

# ===== MODEL =====
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1),
            nn.ReLU(inplace=True),
        )
    def forward(self, x):
        return self.conv(x)

class UNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.down1 = DoubleConv(3, 64)
        self.down2 = DoubleConv(64, 128)
        self.down3 = DoubleConv(128, 256)
        self.down4 = DoubleConv(256, 512)

        self.pool = nn.MaxPool2d(2)

        self.bridge = DoubleConv(512, 1024)

        self.up4 = nn.ConvTranspose2d(1024, 512, 2, stride=2)
        self.conv4 = DoubleConv(1024, 512)

        self.up3 = nn.ConvTranspose2d(512, 256, 2, stride=2)
        self.conv3 = DoubleConv(512, 256)

        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.conv2 = DoubleConv(256, 128)

        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.conv1 = DoubleConv(128, 64)

        self.final = nn.Conv2d(64, 1, 1)

    def forward(self, x):
        d1 = self.down1(x)
        d2 = self.down2(self.pool(d1))
        d3 = self.down3(self.pool(d2))
        d4 = self.down4(self.pool(d3))

        bridge = self.bridge(self.pool(d4))

        u4 = self.up4(bridge)
        u4 = torch.cat([u4, d4], dim=1)
        u4 = self.conv4(u4)

        u3 = self.up3(u4)
        u3 = torch.cat([u3, d3], dim=1)
        u3 = self.conv3(u3)

        u2 = self.up2(u3)
        u2 = torch.cat([u2, d2], dim=1)
        u2 = self.conv2(u2)

        u1 = self.up1(u2)
        u1 = torch.cat([u1, d1], dim=1)
        u1 = self.conv1(u1)

        return torch.sigmoid(self.final(u1))

# ===== LOAD MODEL =====
model = UNet().to(DEVICE)
model.load_state_dict(torch.load(MODEL_PATH, map_location=DEVICE))
model.eval()

print("✅ Model loaded")

# ===== FUNCTION =====
def test_image(image_path):
    img = cv2.imread(image_path)

    if img is None:
        print("❌ Cannot load image")
        return

    original = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = original.shape[:2]

    # ===== preprocess =====
    img_resized = cv2.resize(original, (IMG_SIZE, IMG_SIZE)) / 255.0
    img_input = np.transpose(img_resized, (2, 0, 1))
    img_input = torch.tensor(img_input, dtype=torch.float32).unsqueeze(0).to(DEVICE)

    # ===== prediction =====
    with torch.no_grad():
        pred = model(img_input)   # ✅ FIX (NO sigmoid)

    pred = pred.squeeze().cpu().numpy()

    # ===== DEBUG =====
    print("Pred min:", pred.min(), "max:", pred.max())

    # ===== visualize raw output (VERY IMPORTANT) =====
    plt.imshow(pred, cmap="jet")
    plt.title("Raw Prediction Heatmap")
    plt.colorbar()
    plt.show()

    # ===== threshold =====
    threshold = 0.3   # tune this
    mask = (pred > threshold).astype(np.uint8)

    # resize back
    mask = cv2.resize(mask, (w, h), interpolation=cv2.INTER_NEAREST)

    # ===== DAMAGE % =====
    damage_percent = (mask.sum() / mask.size) * 100
    print(f"🌾 Damage: {damage_percent:.2f}%")

    # ===== overlay =====
    overlay = original.copy()
    overlay[mask == 1] = [255, 0, 0]

    # ===== show =====
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    plt.title("Original")
    plt.imshow(original)
    plt.axis("off")

    plt.subplot(1,3,2)
    plt.title("Mask")
    plt.imshow(mask, cmap="gray")
    plt.axis("off")

    plt.subplot(1,3,3)
    plt.title("Overlay")
    plt.imshow(overlay)
    plt.axis("off")

    plt.show()
```

```python
test_image("img3.jpg")
```
