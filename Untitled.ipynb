```python
import os
import cv2
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import matplotlib.pyplot as plt

# ===== DEVICE (M1 GPU support) =====
DEVICE = "mps" if torch.backends.mps.is_available() else "cpu"

IMG_SIZE = 256
BATCH_SIZE = 4
EPOCHS = 10
LR = 1e-3

TRAIN_IMG_DIR = "dataset_split/train/images"
TRAIN_MASK_DIR = "dataset_split/train/masks"

VAL_IMG_DIR = "dataset_split/val/images"
VAL_MASK_DIR = "dataset_split/val/masks"

print("Using device:", DEVICE)
```

```python
class SegDataset(Dataset):
    def __init__(self, img_dir, mask_dir):
        self.img_dir = img_dir
        self.mask_dir = mask_dir

        self.files = [f for f in os.listdir(img_dir) if not f.startswith(".")]

    def __len__(self):
        return len(self.files)

    def get_mask_path(self, file):
        base = os.path.splitext(file)[0]

        for ext in [".png", ".jpg"]:
            path = os.path.join(self.mask_dir, base + ext)
            if os.path.exists(path):
                return path

        return None

    def __getitem__(self, idx):
        file = self.files[idx]

        img_path = os.path.join(self.img_dir, file)
        mask_path = self.get_mask_path(file)

        # ===== LOAD IMAGE =====
        image = cv2.imread(img_path)

        if image is None:
            print("⚠️ Bad image:", file)
            return self.__getitem__((idx + 1) % len(self))

        image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) / 255.0

        # ===== LOAD MASK =====
        if mask_path is None:
            print("⚠️ Missing mask:", file)
            mask = np.zeros((IMG_SIZE, IMG_SIZE), dtype=np.float32)
        else:
            mask = cv2.imread(mask_path, 0)

            if mask is None:
                print("⚠️ Corrupt mask:", file)
                mask = np.zeros((IMG_SIZE, IMG_SIZE), dtype=np.float32)
            else:
                mask = cv2.resize(mask, (IMG_SIZE, IMG_SIZE))
                mask = (mask > 0).astype(np.float32)

        image = np.transpose(image, (2, 0, 1))

        return torch.tensor(image, dtype=torch.float32), torch.tensor(mask, dtype=torch.float32).unsqueeze(0)
```

```python
train_loader = DataLoader(
    SegDataset(TRAIN_IMG_DIR, TRAIN_MASK_DIR),
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    SegDataset(VAL_IMG_DIR, VAL_MASK_DIR),
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("Train batches:", len(train_loader))
print("Val batches:", len(val_loader))
```

```python
imgs, masks = next(iter(train_loader))

img = imgs[0].permute(1, 2, 0).numpy()
mask = masks[0].squeeze().numpy()

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Image")
plt.imshow(img)

plt.subplot(1,2,2)
plt.title("Mask")
plt.imshow(mask, cmap="gray")

plt.show()
```

```python
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1),
            nn.ReLU(inplace=True),
        )

    def forward(self, x):
        return self.conv(x)

class UNet(nn.Module):
    def __init__(self):
        super().__init__()

        self.down1 = DoubleConv(3, 64)
        self.down2 = DoubleConv(64, 128)
        self.down3 = DoubleConv(128, 256)
        self.down4 = DoubleConv(256, 512)

        self.pool = nn.MaxPool2d(2)

        self.bridge = DoubleConv(512, 1024)

        self.up4 = nn.ConvTranspose2d(1024, 512, 2, stride=2)
        self.conv4 = DoubleConv(1024, 512)

        self.up3 = nn.ConvTranspose2d(512, 256, 2, stride=2)
        self.conv3 = DoubleConv(512, 256)

        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.conv2 = DoubleConv(256, 128)

        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.conv1 = DoubleConv(128, 64)

        self.final = nn.Conv2d(64, 1, 1)

    def forward(self, x):
        d1 = self.down1(x)
        d2 = self.down2(self.pool(d1))
        d3 = self.down3(self.pool(d2))
        d4 = self.down4(self.pool(d3))

        bridge = self.bridge(self.pool(d4))

        u4 = self.up4(bridge)
        u4 = torch.cat([u4, d4], dim=1)
        u4 = self.conv4(u4)

        u3 = self.up3(u4)
        u3 = torch.cat([u3, d3], dim=1)
        u3 = self.conv3(u3)

        u2 = self.up2(u3)
        u2 = torch.cat([u2, d2], dim=1)
        u2 = self.conv2(u2)

        u1 = self.up1(u2)
        u1 = torch.cat([u1, d1], dim=1)
        u1 = self.conv1(u1)

        return torch.sigmoid(self.final(u1))
```

```python
model = UNet().to(DEVICE)

bce = nn.BCELoss()

def dice_loss(pred, target, smooth=1):
    pred = pred.view(-1)
    target = target.view(-1)
    intersection = (pred * target).sum()
    return 1 - (2. * intersection + smooth) / (pred.sum() + target.sum() + smooth)

def loss_fn(pred, target):
    return bce(pred, target) + dice_loss(pred, target)

optimizer = torch.optim.Adam(model.parameters(), lr=LR)
```

```python
best_loss = float("inf")

for epoch in range(EPOCHS):
    model.train()
    train_loss = 0

    for imgs, masks in tqdm(train_loader):
        imgs, masks = imgs.to(DEVICE), masks.to(DEVICE)

        preds = model(imgs)
        loss = loss_fn(preds, masks)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    model.eval()
    val_loss = 0

    with torch.no_grad():
        for imgs, masks in val_loader:
            imgs, masks = imgs.to(DEVICE), masks.to(DEVICE)

            preds = model(imgs)
            loss = loss_fn(preds, masks)
            val_loss += loss.item()

    print(f"Epoch {epoch+1} | Train: {train_loss:.4f} | Val: {val_loss:.4f}")

    if val_loss < best_loss:
        best_loss = val_loss
        torch.save(model.state_dict(), "best_unet.pth")
        print("✅ Model saved")
```
