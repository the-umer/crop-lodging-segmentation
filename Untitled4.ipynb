```python
class CropDamageDataset(Dataset):
    def __init__(self, file_list, image_dir, crop_mask_dir, damage_mask_dir, transform=None):
        self.file_list = file_list
        self.image_dir = image_dir
        self.crop_mask_dir = crop_mask_dir
        self.damage_mask_dir = damage_mask_dir
        self.transform = transform

    def __len__(self):
        return len(self.file_list)

    def __getitem__(self, idx):
        file_name = self.file_list[idx]

        # Load image
        img_path = os.path.join(self.image_dir, file_name)
        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Load masks
        crop_mask = cv2.imread(
            os.path.join(self.crop_mask_dir, file_name),
            cv2.IMREAD_GRAYSCALE
        )
        damage_mask = cv2.imread(
            os.path.join(self.damage_mask_dir, file_name),
            cv2.IMREAD_GRAYSCALE
        )

        # Normalize masks
        crop_mask = (crop_mask > 127).astype(np.float32)
        damage_mask = (damage_mask > 127).astype(np.float32)

        # Restrict damage to crop region
        damage_mask = damage_mask * crop_mask

        if self.transform:
            augmented = self.transform(
                image=image,
                masks=[crop_mask, damage_mask]
            )
            image = augmented["image"]
            crop_mask, damage_mask = augmented["masks"]

        # FIX: Add channel dimension properly (C, H, W)
        crop_mask = crop_mask.unsqueeze(0)
        damage_mask = damage_mask.unsqueeze(0)

        return image, crop_mask, damage_mask
```

```python
# BCE Loss
bce_loss = nn.BCEWithLogitsLoss()

# Dice Loss
def dice_loss(pred, target, smooth=1e-6):
    pred = torch.sigmoid(pred)
    
    intersection = (pred * target).sum()
    union = pred.sum() + target.sum()
    
    dice = (2. * intersection + smooth) / (union + smooth)
    return 1 - dice

# Combined Loss
def combined_loss(pred, target):
    return bce_loss(pred, target) + dice_loss(pred, target)
```

```python
def iou_score(pred, target, threshold=0.5):
    pred = torch.sigmoid(pred)
    pred = (pred > threshold).float()

    intersection = (pred * target).sum()
    union = (pred + target).sum() - intersection

    return (intersection + 1e-6) / (union + 1e-6)


def dice_score(pred, target, threshold=0.5):
    pred = torch.sigmoid(pred)
    pred = (pred > threshold).float()

    intersection = (pred * target).sum()
    return (2 * intersection + 1e-6) / (pred.sum() + target.sum() + 1e-6)
```

```python
from tqdm import tqdm

def train_one_epoch(model, loader, optimizer, device):
    model.train()
    
    total_loss = 0
    total_iou = 0
    total_dice = 0

    loop = tqdm(loader, desc="Training", leave=False)

    for images, _, damage_masks in loop:
        images = images.to(device)
        damage_masks = damage_masks.to(device)

        optimizer.zero_grad()
        outputs = model(images)

        loss = combined_loss(outputs, damage_masks)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        total_iou += iou_score(outputs, damage_masks).item()
        total_dice += dice_score(outputs, damage_masks).item()

        # Update progress bar
        loop.set_postfix(loss=loss.item())

    return (
        total_loss / len(loader),
        total_iou / len(loader),
        total_dice / len(loader)
    )


def validate_one_epoch(model, loader, device):
    model.eval()
    
    total_loss = 0
    total_iou = 0
    total_dice = 0

    loop = tqdm(loader, desc="Validation", leave=False)

    with torch.no_grad():
        for images, _, damage_masks in loop:
            images = images.to(device)
            damage_masks = damage_masks.to(device)

            outputs = model(images)
            loss = combined_loss(outputs, damage_masks)

            total_loss += loss.item()
            total_iou += iou_score(outputs, damage_masks).item()
            total_dice += dice_score(outputs, damage_masks).item()

            loop.set_postfix(loss=loss.item())

    return (
        total_loss / len(loader),
        total_iou / len(loader),
        total_dice / len(loader)
    )
```

```python
num_epochs = 15

best_val_iou = 0

train_losses, val_losses = [], []
train_ious, val_ious = [], []
train_dices, val_dices = [], []

for epoch in range(num_epochs):
    print(f"\nEpoch {epoch+1}/{num_epochs}")

    train_loss, train_iou, train_dice = train_one_epoch(
        model, train_loader, optimizer, device
    )

    val_loss, val_iou, val_dice = validate_one_epoch(
        model, val_loader, device
    )

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    train_ious.append(train_iou)
    val_ious.append(val_iou)

    train_dices.append(train_dice)
    val_dices.append(val_dice)

    print(f"Train Loss: {train_loss:.4f} | IoU: {train_iou:.4f} | Dice: {train_dice:.4f}")
    print(f"Val   Loss: {val_loss:.4f} | IoU: {val_iou:.4f} | Dice: {val_dice:.4f}")

    # Save best model
    if val_iou > best_val_iou:
        best_val_iou = val_iou
        torch.save(model.state_dict(), "best_unet_model.pth")
        print("✅ Best model saved!")
```

```python
def evaluate_model(model, loader, device):
    model.eval()

    total_iou = 0
    total_dice = 0
    total_mae = 0

    with torch.no_grad():
        for images, crop_masks, damage_masks in loader:
            images = images.to(device)
            crop_masks = crop_masks.to(device)
            damage_masks = damage_masks.to(device)

            outputs = model(images)
            preds = torch.sigmoid(outputs)

            # Binarize
            preds_bin = (preds > 0.5).float()

            # IoU & Dice
            total_iou += iou_score(outputs, damage_masks).item()
            total_dice += dice_score(outputs, damage_masks).item()

            # MAE (only inside crop region)
            mae = torch.abs(preds_bin - damage_masks) * crop_masks
            total_mae += mae.mean().item()

    return (
        total_iou / len(loader),
        total_dice / len(loader),
        total_mae / len(loader)
    )
```

```python
def compute_damage_percentage(pred_mask, crop_mask):
    """
    pred_mask: predicted binary mask (1, H, W)
    crop_mask: crop region mask (1, H, W)
    """

    # Only consider crop region
    damage_pixels = (pred_mask * crop_mask).sum()
    total_crop_pixels = crop_mask.sum()

    if total_crop_pixels == 0:
        return 0.0

    return (damage_pixels / total_crop_pixels).item() * 100
```

```python
def visualize_prediction_with_errors(model, dataset, idx, device):
    model.eval()

    image, crop_mask, damage_mask = dataset[idx]

    # Prepare input
    input_image = image.unsqueeze(0).to(device)
    crop_mask = crop_mask.to(device)
    gt_mask = damage_mask.to(device)

    with torch.no_grad():
        output = model(input_image)
        pred = torch.sigmoid(output)
        pred_bin = (pred > 0.5).float()

    # Convert to numpy
    image_np = image.permute(1, 2, 0).cpu().numpy()
    image_np = denormalize(image_np)

    crop_mask_np = crop_mask.squeeze().cpu().numpy()
    gt_mask_np = gt_mask.squeeze().cpu().numpy()
    pred_mask_np = pred_bin.squeeze().cpu().numpy()

    # Compute TP, FP, FN
    tp = (pred_mask_np == 1) & (gt_mask_np == 1)
    fp = (pred_mask_np == 1) & (gt_mask_np == 0)
    fn = (pred_mask_np == 0) & (gt_mask_np == 1)

    # Create overlay
    overlay = image_np.copy()

    # Apply colors
    overlay[tp] = [0, 1, 0]   # Green → correct
    overlay[fp] = [1, 0, 0]   # Red → false positive
    overlay[fn] = [0, 0, 1]   # Blue → missed

    # Blend
    blended = 0.6 * image_np + 0.4 * overlay

    # Compute damage %
    damage_percent = compute_damage_percentage(pred_bin, crop_mask)

    # Plot
    plt.figure(figsize=(18, 5))

    plt.subplot(1, 4, 1)
    plt.imshow(image_np)
    plt.title("Original")
    plt.axis("off")

    plt.subplot(1, 4, 2)
    plt.imshow(gt_mask_np, cmap="gray")
    plt.title("Ground Truth")
    plt.axis("off")

    plt.subplot(1, 4, 3)
    plt.imshow(pred_mask_np, cmap="gray")
    plt.title("Prediction")
    plt.axis("off")

    plt.subplot(1, 4, 4)
    plt.imshow(blended)
    plt.title(f"Overlay (Damage: {damage_percent:.2f}%)")
    plt.axis("off")

    plt.show()
```

```python
def visualize_psp_result(model, dataset, idx, device):
    model.eval()

    image, crop_mask, damage_mask = dataset[idx]

    input_image = image.unsqueeze(0).to(device)
    crop_mask = crop_mask.to(device)
    gt_mask = damage_mask.to(device)

    with torch.no_grad():
        output = model(input_image)
        pred = torch.sigmoid(output)
        pred_bin = (pred > 0.5).float()

    # Convert to numpy
    image_np = denormalize(image.permute(1,2,0).cpu().numpy())
    gt_mask_np = gt_mask.squeeze().cpu().numpy()
    pred_mask_np = pred_bin.squeeze().cpu().numpy()

    # TP / FP / FN
    tp = (pred_mask_np == 1) & (gt_mask_np == 1)
    fp = (pred_mask_np == 1) & (gt_mask_np == 0)
    fn = (pred_mask_np == 0) & (gt_mask_np == 1)

    overlay = image_np.copy()
    overlay[tp] = [0,1,0]   # green
    overlay[fp] = [1,0,0]   # red
    overlay[fn] = [0,0,1]   # blue

    blended = 0.6 * image_np + 0.4 * overlay

    damage_percent = compute_damage_percentage(pred_bin, crop_mask)

    # Plot
    plt.figure(figsize=(20,5))

    plt.subplot(1,4,1)
    plt.imshow(image_np)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1,4,2)
    plt.imshow(gt_mask_np, cmap="gray")
    plt.title("Ground Truth")
    plt.axis("off")

    plt.subplot(1,4,3)
    plt.imshow(pred_mask_np, cmap="gray")
    plt.title("Prediction")
    plt.axis("off")

    plt.subplot(1,4,4)
    plt.imshow(blended)
    plt.title(f"Overlay\nDamage: {damage_percent:.2f}%")
    plt.axis("off")

    plt.show()
```

```python
for i in range(5):
    visualize_psp_result(model_psp, test_dataset, i, device)
```

```python
def save_psp_results(model, dataset, device, save_dir):
    os.makedirs(save_dir, exist_ok=True)

    for idx in range(len(dataset)):
        image, crop_mask, damage_mask = dataset[idx]

        input_image = image.unsqueeze(0).to(device)
        crop_mask = crop_mask.to(device)
        gt_mask = damage_mask.to(device)

        with torch.no_grad():
            output = model(input_image)
            pred = torch.sigmoid(output)
            pred_bin = (pred > 0.5).float()

        image_np = denormalize(image.permute(1,2,0).cpu().numpy())
        gt_mask_np = damage_mask.squeeze().cpu().numpy()
        pred_mask_np = pred_bin.squeeze().cpu().numpy()

        tp = (pred_mask_np == 1) & (gt_mask_np == 1)
        fp = (pred_mask_np == 1) & (gt_mask_np == 0)
        fn = (pred_mask_np == 0) & (gt_mask_np == 1)

        overlay = image_np.copy()
        overlay[tp] = [0,1,0]
        overlay[fp] = [1,0,0]
        overlay[fn] = [0,0,1]

        blended = 0.6 * image_np + 0.4 * overlay

        damage_percent = compute_damage_percentage(pred_bin, crop_mask)

        save_path = os.path.join(save_dir, f"sample_{idx}.png")

        plt.figure(figsize=(5,5))
        plt.imshow(blended)
        plt.title(f"{damage_percent:.2f}%")
        plt.axis("off")
        plt.savefig(save_path, bbox_inches='tight')
        plt.close()

    print("Saved results in:", save_dir)
```
