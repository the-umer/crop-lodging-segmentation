```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from glob import glob

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

# Device setup for M1
DEVICE = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", DEVICE)

# Paths
BASE_PATH = "dataset_split"

TRAIN_IMG_PATH = os.path.join(BASE_PATH, "train/images")
TRAIN_MASK_PATH = os.path.join(BASE_PATH, "train/masks")

VAL_IMG_PATH = os.path.join(BASE_PATH, "val/images")
VAL_MASK_PATH = os.path.join(BASE_PATH, "val/masks")

IMG_SIZE = 256
BATCH_SIZE = 8
```

```python
def get_mask_path(img_path, mask_dir):
    name = os.path.splitext(os.path.basename(img_path))[0]
    
    mask_jpg = os.path.join(mask_dir, name + ".jpg")
    mask_png = os.path.join(mask_dir, name + ".png")
    
    if os.path.exists(mask_jpg):
        return mask_jpg
    elif os.path.exists(mask_png):
        return mask_png
    else:
        return None

train_images = sorted(glob(TRAIN_IMG_PATH + "/*.jpg"))
val_images = sorted(glob(VAL_IMG_PATH + "/*.jpg"))

print("Train images:", len(train_images))
print("Val images:", len(val_images))

# Check matching
missing_masks = 0

for img_path in train_images:
    if get_mask_path(img_path, TRAIN_MASK_PATH) is None:
        missing_masks += 1

print("Missing train masks:", missing_masks)

missing_masks_val = 0
for img_path in val_images:
    if get_mask_path(img_path, VAL_MASK_PATH) is None:
        missing_masks_val += 1

print("Missing val masks:", missing_masks_val)
```

```python
class CropDataset(Dataset):
    def __init__(self, image_paths, mask_dir):
        self.image_paths = image_paths
        self.mask_dir = mask_dir

    def get_mask_path(self, img_path):
        name = os.path.splitext(os.path.basename(img_path))[0]
        
        mask_jpg = os.path.join(self.mask_dir, name + ".jpg")
        mask_png = os.path.join(self.mask_dir, name + ".png")
        
        if os.path.exists(mask_jpg):
            return mask_jpg
        elif os.path.exists(mask_png):
            return mask_png
        else:
            return None

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        mask_path = self.get_mask_path(img_path)

        # ---- Read image ----
        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
        image = image / 255.0
        image = np.transpose(image, (2, 0, 1))  # (C, H, W)

        # ---- Read mask ----
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        mask = cv2.resize(mask, (IMG_SIZE, IMG_SIZE))

        # ---- IMPORTANT: convert to binary ----
        mask = (mask > 127).astype(np.float32)

        mask = np.expand_dims(mask, axis=0)  # (1, H, W)

        return torch.tensor(image, dtype=torch.float32), torch.tensor(mask, dtype=torch.float32)
```

```python
train_dataset = CropDataset(train_images, TRAIN_MASK_PATH)
val_dataset = CropDataset(val_images, VAL_MASK_PATH)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)

print("Train batches:", len(train_loader))
print("Val batches:", len(val_loader))
```

```python
def visualize_sample(dataset, idx=0):
    image, mask = dataset[idx]

    image = image.numpy().transpose(1, 2, 0)
    mask = mask.numpy().squeeze()

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.title("Image")
    plt.imshow(image)

    plt.subplot(1, 3, 2)
    plt.title("Mask")
    plt.imshow(mask, cmap='gray')

    plt.subplot(1, 3, 3)
    plt.title("Overlay")
    overlay = image.copy()
    overlay[mask == 1] = [1, 0, 0]  # red highlight
    plt.imshow(overlay)

    plt.show()

# Show few samples
for i in range(3):
    visualize_sample(train_dataset, i)
```

```python
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super(DoubleConv, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),

            nn.Conv2d(out_ch, out_ch, 3, padding=1),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
        )

    def forward(self, x):
        return self.conv(x)


class UNet(nn.Module):
    def __init__(self):
        super(UNet, self).__init__()

        self.down1 = DoubleConv(3, 64)
        self.pool1 = nn.MaxPool2d(2)

        self.down2 = DoubleConv(64, 128)
        self.pool2 = nn.MaxPool2d(2)

        self.down3 = DoubleConv(128, 256)
        self.pool3 = nn.MaxPool2d(2)

        self.bridge = DoubleConv(256, 512)

        self.up3 = nn.ConvTranspose2d(512, 256, 2, stride=2)
        self.conv3 = DoubleConv(512, 256)

        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.conv2 = DoubleConv(256, 128)

        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.conv1 = DoubleConv(128, 64)

        self.out = nn.Conv2d(64, 1, kernel_size=1)

    def forward(self, x):
        d1 = self.down1(x)
        d2 = self.down2(self.pool1(d1))
        d3 = self.down3(self.pool2(d2))

        bridge = self.bridge(self.pool3(d3))

        u3 = self.up3(bridge)
        u3 = torch.cat([u3, d3], dim=1)
        u3 = self.conv3(u3)

        u2 = self.up2(u3)
        u2 = torch.cat([u2, d2], dim=1)
        u2 = self.conv2(u2)

        u1 = self.up1(u2)
        u1 = torch.cat([u1, d1], dim=1)
        u1 = self.conv1(u1)

        out = self.out(u1)
        return out
```

```python
model = UNet().to(DEVICE)
print("Model loaded on:", DEVICE)
```

```python
bce_loss = nn.BCEWithLogitsLoss()

def dice_loss(pred, target, smooth=1):
    pred = torch.sigmoid(pred)
    pred = pred.view(-1)
    target = target.view(-1)

    intersection = (pred * target).sum()
    dice = (2. * intersection + smooth) / (pred.sum() + target.sum() + smooth)

    return 1 - dice


def combined_loss(pred, target):
    return bce_loss(pred, target) + dice_loss(pred, target)
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
```

```python
EPOCHS = 15
best_val_loss = float("inf")

train_losses = []
val_losses = []

for epoch in range(EPOCHS):
    # ===== TRAIN =====
    model.train()
    train_loss = 0

    for images, masks in train_loader:
        images = images.to(DEVICE)
        masks = masks.to(DEVICE)

        optimizer.zero_grad()

        outputs = model(images)
        loss = combined_loss(outputs, masks)

        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    train_loss /= len(train_loader)
    train_losses.append(train_loss)

    # ===== VALIDATION =====
    model.eval()
    val_loss = 0

    with torch.no_grad():
        for images, masks in val_loader:
            images = images.to(DEVICE)
            masks = masks.to(DEVICE)

            outputs = model(images)
            loss = combined_loss(outputs, masks)

            val_loss += loss.item()

    val_loss /= len(val_loader)
    val_losses.append(val_loss)

    print(f"Epoch [{epoch+1}/{EPOCHS}] | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")

    # ===== SAVE BEST MODEL =====
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        torch.save(model.state_dict(), "best_model.pth")
        print("✅ Model Saved!")
```

```python
def calculate_metrics(pred, target):
    pred = torch.sigmoid(pred)
    pred = (pred > 0.5).float()

    pred = pred.view(-1)
    target = target.view(-1)

    # Intersection
    intersection = (pred * target).sum()

    # IoU
    union = pred.sum() + target.sum() - intersection
    iou = (intersection + 1e-6) / (union + 1e-6)

    # Dice
    dice = (2 * intersection + 1e-6) / (pred.sum() + target.sum() + 1e-6)

    # Accuracy
    correct = (pred == target).sum()
    acc = correct / target.numel()

    return iou.item(), dice.item(), acc.item()
```

```python
iou_total = 0
dice_total = 0
acc_total = 0

model.eval()

with torch.no_grad():
    for images, masks in val_loader:
        images = images.to(DEVICE)
        masks = masks.to(DEVICE)

        outputs = model(images)

        iou, dice, acc = calculate_metrics(outputs, masks)

        iou_total += iou
        dice_total += dice
        acc_total += acc

n = len(val_loader)

print(f"Mean IoU: {iou_total/n:.4f}")
print(f"Mean Dice: {dice_total/n:.4f}")
print(f"Mean Accuracy: {acc_total/n:.4f}")
```

```python
def predict_damage(image_path):
    # ---- Read image ----
    image = cv2.imread(image_path)
    original = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    img = cv2.resize(original, (IMG_SIZE, IMG_SIZE))
    img = img / 255.0
    img = np.transpose(img, (2, 0, 1))
    img = np.expand_dims(img, axis=0)

    img_tensor = torch.tensor(img, dtype=torch.float32).to(DEVICE)

    # ---- Prediction ----
    with torch.no_grad():
        output = model(img_tensor)
        pred = torch.sigmoid(output).cpu().numpy().squeeze()

    # ---- Binary mask ----
    pred_mask = (pred > 0.5).astype(np.uint8)

    # ---- Damage % ----
    damaged_pixels = np.sum(pred_mask)
    total_pixels = pred_mask.size

    damage_percent = (damaged_pixels / total_pixels) * 100

    print(f"🌾 Damage Percentage: {damage_percent:.2f}%")

    # ---- Visualization ----
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.title("Original")
    plt.imshow(original)

    plt.subplot(1, 3, 2)
    plt.title("Predicted Mask")
    plt.imshow(pred_mask, cmap='gray')

    plt.subplot(1, 3, 3)
    plt.title("Overlay")
    overlay = original.copy()
    overlay = cv2.resize(overlay, (IMG_SIZE, IMG_SIZE))
    overlay[pred_mask == 1] = [255, 0, 0]
    plt.imshow(overlay)

    plt.show()

    return damage_percent
```
